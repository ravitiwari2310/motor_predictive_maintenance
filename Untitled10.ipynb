```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
output_dir = "pmsm_thermal_twin/eda_output"

# Load dataset
df = pd.read_csv("pmsm_thermal_twin/data/raw/measures_v2.csv")
df['i_mag_sq'] = df['i_d']**2 + df['i_q']**2

# Step 1: Compute Cross-Correlation Function (CCF) for a representative profile (Profile 6)
profile_sample = df[df['profile_id'] == 6].reset_index(drop=True)

lags = [1, 2, 5, 10, 20, 30, 60, 120]  # Corresponding to 0.5s, 1s, 2.5s, 5s, 10s, 15s, 30s, 60s
ccf_results = []
acf_results = []

temp_series = profile_sample['stator_winding']
current_series = profile_sample['i_mag_sq']

for lag in lags:
    # Autocorrelation of temperature
    acf_val = temp_series.autocorr(lag=lag)
    acf_results.append({'lag_steps': lag, 'time_seconds': lag * 0.5, 'ACF': acf_val})

    # Cross-correlation: current(t - lag) vs temp(t)
    shifted_current = current_series.shift(lag)
    valid_idx = shifted_current.dropna().index
    ccf_val = np.corrcoef(shifted_current.dropna(), temp_series.loc[valid_idx])[0, 1]
    ccf_results.append({'lag_steps': lag, 'time_seconds': lag * 0.5, 'CCF_Current_Temp': ccf_val})

df_acf = pd.DataFrame(acf_results)
df_ccf = pd.DataFrame(ccf_results)

# Step 2: Plot ACF and CCF
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=df_acf, x='time_seconds', y='ACF', ax=ax1, palette='Blues_d')
ax1.set_title('Stator Winding Temperature Autocorrelation (ACF)', fontweight='bold')
ax1.set_xlabel('Lag Time (Seconds)')
ax1.set_ylabel('Autocorrelation')

sns.barplot(data=df_ccf, x='time_seconds', y='CCF_Current_Temp', ax=ax2, palette='Oranges_d')
ax2.set_title('Current Loss Proxy vs. Temp Cross-Correlation (CCF)', fontweight='bold')
ax2.set_xlabel('Lag Time (Seconds)')
ax2.set_ylabel('Cross-Correlation')

plt.tight_layout()
fig_path = os.path.join(output_dir, "fig5_acf_ccf_analysis.png")
plt.savefig(fig_path, dpi=300)
plt.show()

print("=== AUTOCORRELATION TABLE (TEMPERATURE SELF-DEPENDENCY) ===")
print(df_acf.to_string(index=False))

print("\n=== CROSS-CORRELATION TABLE (CURRENT LOSS TO TEMP LAG) ===")
print(df_ccf.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import os

# Step 1: Load raw data
raw_path = "pmsm_thermal_twin/data/raw/measures_v2.csv"
df = pd.read_csv(raw_path)

FORECAST_STEPS = 120  # 60 seconds forecast (at 2 Hz sampling)

print("Engineering features across profiles...")

# Step 2: Feature creation function grouped by profile_id
def engineer_profile_features(group):
    # Targets
    group['stator_winding_future_60s'] = group['stator_winding'].shift(-FORECAST_STEPS)
    group['temp_rise_60s'] = group['stator_winding_future_60s'] - group['stator_winding']

    # Loss & Electrical Proxies
    group['i_mag_sq'] = group['i_d']**2 + group['i_q']**2         # Copper loss proxy
    group['u_mag_sq'] = group['u_d']**2 + group['u_q']**2         # Voltage magnitude proxy
    group['speed_sq'] = group['motor_speed']**2                   # Core loss proxy
    group['mech_power_proxy'] = group['motor_speed'] * group['torque'] # Shaft power proxy

    # Thermal Gradients
    group['winding_tooth_diff'] = group['stator_winding'] - group['stator_tooth']
    group['winding_yoke_diff'] = group['stator_winding'] - group['stator_yoke']
    group['coolant_yoke_diff'] = group['stator_yoke'] - group['coolant']

    # Lags (Past observations only)
    group['winding_lag_10s'] = group['stator_winding'].shift(20)  # 20 steps = 10s
    group['winding_lag_30s'] = group['stator_winding'].shift(60)  # 60 steps = 30s

    # Slopes & Loss Accumulation
    group['temp_slope_10s'] = (group['stator_winding'] - group['winding_lag_10s']) / 10.0
    group['i_mag_sq_roll_mean_30s'] = group['i_mag_sq'].rolling(window=60, min_periods=1).mean()

    return group

# Apply group-wise transformation
df_featured = df.groupby('profile_id', group_keys=False).apply(engineer_profile_features)

# Step 3: Clean boundary NaNs (resulting from past lags and future targets)
initial_shape = df_featured.shape
df_clean = df_featured.dropna().reset_index(drop=True)
final_shape = df_clean.shape

# Step 4: Save clean featured dataset to processed folder
processed_dir = "pmsm_thermal_twin/data/processed"
os.makedirs(processed_dir, exist_ok=True)
output_path = os.path.join(processed_dir, "pmsm_featured.csv")
df_clean.to_csv(output_path, index=False)

print("\n=== FEATURE ENGINEERING SUMMARY ===")
print(f"Original Telemetry Rows: {initial_shape[0]}")
print(f"Cleaned Telemetry Rows:  {final_shape[0]} ({(final_shape[0]/initial_shape[0])*100:.2f}% retained)")
print(f"Total Feature Columns:  {final_shape[1]}")
print(f"Dataset Saved To:       {output_path}")

print("\n=== ENGINEERED FEATURE COLUMNS ===")
for i, col in enumerate(df_clean.columns, 1):
    print(f"{i:02d}. {col}")
```

```python
import pandas as pd
import numpy as np
import os

# Load featured dataset
processed_path = "pmsm_thermal_twin/data/processed/pmsm_featured.csv"
df = pd.read_csv(processed_path)

# Get unique profiles
unique_profiles = df['profile_id'].unique()
np.random.seed(42)  # For strict reproducibility
np.random.shuffle(unique_profiles)

# Split ratios: 70% Train, 15% Validation, 15% Test
n_total = len(unique_profiles)
n_train = int(0.70 * n_total)
n_val = int(0.15 * n_total)

train_profiles = unique_profiles[:n_train]
val_profiles = unique_profiles[n_train:n_train + n_val]
test_profiles = unique_profiles[n_train + n_val:]

# Subset dataframes
train_df = df[df['profile_id'].isin(train_profiles)].reset_index(drop=True)
val_df = df[df['profile_id'].isin(val_profiles)].reset_index(drop=True)
test_df = df[df['profile_id'].isin(test_profiles)].reset_index(drop=True)

print("=== PROFILE SPLIT SUMMARY ===")
print(f"Total Profiles: {n_total}")
print(f"Train Profiles ({len(train_profiles)}): {sorted(train_profiles)}")
print(f"Val Profiles   ({len(val_profiles)}):  {sorted(val_profiles)}")
print(f"Test Profiles  ({len(test_profiles)}):  {sorted(test_profiles)}")

print("\n=== ROW DISTRIBUTION ===")
print(f"Train Rows: {len(train_df):,} ({(len(train_df)/len(df))*100:.2f}%)")
print(f"Val Rows:   {len(val_df):,} ({(len(val_df)/len(df))*100:.2f}%)")
print(f"Test Rows:  {len(test_df):,} ({(len(test_df)/len(df))*100:.2f}%)")

print("\n=== TARGET DISTRIBUTION CHECK (STATOR WINDING FUTURE 60S) ===")
print(f"Train Temp Mean: {train_df['stator_winding_future_60s'].mean():.2f}°C, Std: {train_df['stator_winding_future_60s'].std():.2f}°C")
print(f"Val Temp Mean:   {val_df['stator_winding_future_60s'].mean():.2f}°C, Std: {val_df['stator_winding_future_60s'].std():.2f}°C")
print(f"Test Temp Mean:  {test_df['stator_winding_future_60s'].mean():.2f}°C, Std: {test_df['stator_winding_future_60s'].std():.2f}°C")

# Save split indices / metadata to processed folder for strict pipeline consistency
np.savez(
    "pmsm_thermal_twin/data/processed/split_profiles.npz",
    train_profiles=train_profiles,
    val_profiles=val_profiles,
    test_profiles=test_profiles
)
print("\nProfile split saved to: pmsm_thermal_twin/data/processed/split_profiles.npz")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

# 1. Load featured dataset and profile splits
df = pd.read_csv("pmsm_thermal_twin/data/processed/pmsm_featured.csv")
splits = np.load("pmsm_thermal_twin/data/processed/split_profiles.npz")

train_profiles = splits['train_profiles']
val_profiles = splits['val_profiles']

train_df = df[df['profile_id'].isin(train_profiles)].reset_index(drop=True)
val_df = df[df['profile_id'].isin(val_profiles)].reset_index(drop=True)

# 2. Separate features and target
ignore_cols = ['profile_id', 'stator_winding_future_60s', 'temp_rise_60s']
feature_cols = [c for c in df.columns if c not in ignore_cols]
target_col = 'stator_winding_future_60s'

X_train, y_train = train_df[feature_cols], train_df[target_col]
X_val, y_val = val_df[feature_cols], val_df[target_col]

# 3. Baseline 1: Naive Persistence Model (T_future = T_current)
y_pred_naive = val_df['stator_winding']

mae_naive = mean_absolute_error(y_val, y_pred_naive)
rmse_naive = root_mean_squared_error(y_val, y_pred_naive)
r2_naive = r2_score(y_val, y_pred_naive)

# 4. Baseline 2: Ridge Linear Regression
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

ridge = Ridge(alpha=100.0)
ridge.fit(X_train_scaled, y_train)
y_pred_ridge = ridge.predict(X_val_scaled)

mae_ridge = mean_absolute_error(y_val, y_pred_ridge)
rmse_ridge = root_mean_squared_error(y_val, y_pred_ridge)
r2_ridge = r2_score(y_val, y_pred_ridge)

# 5. Print Baseline Performance Comparison
results_df = pd.DataFrame({
    'Model': ['Naive Persistence', 'Ridge Regression'],
    'MAE (°C)': [mae_naive, mae_ridge],
    'RMSE (°C)': [rmse_naive, rmse_ridge],
    'R² Score': [r2_naive, r2_ridge]
})

print("=== BASELINE MODEL PERFORMANCE (VALIDATION SET) ===")
print(results_df.to_string(index=False))

print("\n=== TOP 5 MOST IMPORTANT RIDGE COEFFICIENTS ===")
coef_series = pd.Series(ridge.coef_, index=feature_cols).abs().sort_values(ascending=False)
print(coef_series.head(5))
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import xgboost as xgb
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

# 1. Load featured dataset and profile splits
df = pd.read_csv("pmsm_thermal_twin/data/processed/pmsm_featured.csv")
splits = np.load("pmsm_thermal_twin/data/processed/split_profiles.npz")

train_profiles = splits['train_profiles']
val_profiles = splits['val_profiles']

train_df = df[df['profile_id'].isin(train_profiles)].reset_index(drop=True)
val_df = df[df['profile_id'].isin(val_profiles)].reset_index(drop=True)

ignore_cols = ['profile_id', 'stator_winding_future_60s', 'temp_rise_60s']
feature_cols = [c for c in df.columns if c not in ignore_cols]
target_col = 'stator_winding_future_60s'

X_train, y_train = train_df[feature_cols], train_df[target_col]
X_val, y_val = val_df[feature_cols], val_df[target_col]

# 2. LightGBM Model
print("Training LightGBM Regressor...")
lgb_model = lgb.LGBMRegressor(
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    random_state=42,
    n_jobs=-1
)
lgb_model.fit(X_train, y_train)
y_pred_lgb = lgb_model.predict(X_val)

mae_lgb = mean_absolute_error(y_val, y_pred_lgb)
rmse_lgb = root_mean_squared_error(y_val, y_pred_lgb)
r2_lgb = r2_score(y_val, y_pred_lgb)

# 3. XGBoost Model
print("Training XGBoost Regressor...")
xgb_model = xgb.XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    n_jobs=-1,
    tree_method='hist'
)
xgb_model.fit(X_train, y_train)
y_pred_xgb = xgb_model.predict(X_val)

mae_xgb = mean_absolute_error(y_val, y_pred_xgb)
rmse_xgb = root_mean_squared_error(y_val, y_pred_xgb)
r2_xgb = r2_score(y_val, y_pred_xgb)

# 4. Consolidate Validation Results
val_comparison = pd.DataFrame({
    'Model': ['Naive Persistence', 'Ridge Regression', 'LightGBM', 'XGBoost'],
    'MAE (°C)': [3.4715, 2.4050, mae_lgb, mae_xgb],
    'RMSE (°C)': [5.1571, 3.7296, rmse_lgb, rmse_xgb],
    'R² Score': [0.9566, 0.9773, r2_lgb, r2_xgb]
})

print("\n=== MODEL PERFORMANCE COMPARISON (VALIDATION SET) ===")
print(val_comparison.to_string(index=False))

# 5. Extract Feature Importances from LightGBM
print("\n=== LIGHTGBM TOP 10 FEATURE IMPORTANCES ===")
importances = pd.Series(lgb_model.feature_importances_, index=feature_cols).sort_values(ascending=False)
print(importances.head(10))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
output_dir = "pmsm_thermal_twin/eda_output"

# 1. Load data and profile splits
df = pd.read_csv("pmsm_thermal_twin/data/processed/pmsm_featured.csv")
splits = np.load("pmsm_thermal_twin/data/processed/split_profiles.npz")

train_df = df[df['profile_id'].isin(splits['train_profiles'])].reset_index(drop=True)
val_df = df[df['profile_id'].isin(splits['val_profiles'])].reset_index(drop=True)

ignore_cols = ['profile_id', 'stator_winding_future_60s', 'temp_rise_60s']
feature_cols = [c for c in df.columns if c not in ignore_cols]
target_col = 'stator_winding_future_60s'

X_train, y_train = train_df[feature_cols], train_df[target_col]
X_val, y_val = val_df[feature_cols], val_df[target_col]

# 2. Fit models and get predictions
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_val_s = scaler.transform(X_val)

ridge = Ridge(alpha=100.0).fit(X_train_s, y_train)
y_pred_ridge = ridge.predict(X_val_s)

lgb_model = lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=31, random_state=42, n_jobs=-1)
lgb_model.fit(X_train, y_train)
y_pred_lgb = lgb_model.predict(X_val)

# 3. Calculate Diagnostic Residuals
val_df['err_ridge'] = y_val - y_pred_ridge
val_df['err_lgb'] = y_val - y_pred_lgb
val_df['abs_err_ridge'] = val_df['err_ridge'].abs()
val_df['abs_err_lgb'] = val_df['err_lgb'].abs()

# Categorize Operating Thermal States
val_df['thermal_state'] = np.where(val_df['temp_slope_10s'] > 0.05, 'Rapid Heating',
                          np.where(val_df['temp_slope_10s'] < -0.05, 'Cooling', 'Steady State'))

# 4. Error Breakdown by Thermal Operating State
error_by_state = val_df.groupby('thermal_state').agg(
    row_count=('stator_winding', 'count'),
    ridge_mae=('abs_err_ridge', 'mean'),
    lgb_mae=('abs_err_lgb', 'mean'),
    ridge_max_err=('abs_err_ridge', 'max'),
    lgb_max_err=('abs_err_lgb', 'max')
).reset_index()

print("=== ERROR ANALYSIS BY THERMAL OPERATING REGIME ===")
print(error_by_state.to_string(index=False))

# 5. Error Breakdown by Profile ID
profile_errors = val_df.groupby('profile_id').agg(
    avg_temp=('stator_winding_future_60s', 'mean'),
    ridge_mae=('abs_err_ridge', 'mean'),
    lgb_mae=('abs_err_lgb', 'mean')
).reset_index()

print("\n=== TOP 5 HIGHEST ERROR VALIDATION PROFILES ===")
print(profile_errors.sort_values(by='ridge_mae', ascending=False).head(5).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
import os

sns.set_theme(style="whitegrid")
output_dir = "pmsm_thermal_twin/eda_output"

# 1. Load data and profile splits
df = pd.read_csv("pmsm_thermal_twin/data/processed/pmsm_featured.csv")
splits = np.load("pmsm_thermal_twin/data/processed/split_profiles.npz")

train_df = df[df['profile_id'].isin(splits['train_profiles'])].reset_index(drop=True)
test_df = df[df['profile_id'].isin(splits['test_profiles'])].reset_index(drop=True)

ignore_cols = ['profile_id', 'stator_winding_future_60s', 'temp_rise_60s']
feature_cols = [c for c in df.columns if c not in ignore_cols]
target_col = 'stator_winding_future_60s'

X_train, y_train = train_df[feature_cols], train_df[target_col]
X_test, y_test = test_df[feature_cols], test_df[target_col]

# 2. Train Production Digital Twin (Ridge Regression)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

model = Ridge(alpha=100.0)
model.fit(X_train_s, y_train)
y_pred_test = model.predict(X_test_s)

# 3. Calculate Final Test Performance
mae_test = mean_absolute_error(y_test, y_pred_test)
rmse_test = root_mean_squared_error(y_test, y_pred_test)
r2_test = r2_score(y_test, y_pred_test)

# 4. Anomaly Detection Engine Logic
test_df['pred_winding_60s'] = y_pred_test
test_df['abs_residual'] = (test_df['stator_winding_future_60s'] - test_df['pred_winding_60s']).abs()

ANOMALY_THRESHOLD = 10.0  # °C deviation threshold
test_df['is_anomaly'] = test_df['abs_residual'] > ANOMALY_THRESHOLD

total_test_points = len(test_df)
anomaly_points = test_df['is_anomaly'].sum()

print("=== FINAL HOLDOUT TEST SET PERFORMANCE ===")
print(f"Test MAE:   {mae_test:.4f} °C")
print(f"Test RMSE:  {rmse_test:.4f} °C")
print(f"Test R²:    {r2_test:.4f}")

print("\n=== THERMAL ANOMALY DETECTION SUMMARY ===")
print(f"Anomaly Residual Threshold: > {ANOMALY_THRESHOLD} °C")
print(f"Total Telemetry Sample Points: {total_test_points:,}")
print(f"Flagged Anomaly Points:      {anomaly_points:,} ({(anomaly_points/total_test_points)*100:.2f}%)")

# 5. Plot Digital Twin Prediction vs Actual on a Sample Holdout Test Profile (Profile 60)
sample_test_profile = test_df[test_df['profile_id'] == 60].reset_index(drop=True)

plt.figure(figsize=(14, 6))
plt.plot(sample_test_profile.index, sample_test_profile['stator_winding_future_60s'],
         label='Actual Temp (+60s)', color='black', linewidth=1.8)
plt.plot(sample_test_profile.index, sample_test_profile['pred_winding_60s'],
         label='Digital Twin Forecast (+60s)', color='crimson', linestyle='--', linewidth=1.5)

# Highlight anomaly zones if any exist in profile
anomalies_sample = sample_test_profile[sample_test_profile['is_anomaly']]
if not anomalies_sample.empty:
    plt.scatter(anomalies_sample.index, anomalies_sample['stator_winding_future_60s'],
                color='gold', edgecolor='black', s=40, label='Thermal Anomaly Flag (>10°C Error)', zorder=5)

plt.title('Digital Twin 60-Second Stator Winding Forecast vs Actuals — Test Profile 60', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Time Step (0.5s intervals)', fontsize=11)
plt.ylabel('Temperature (°C)', fontsize=11)
plt.legend(loc='upper left')
plt.tight_layout()

fig_path = os.path.join(output_dir, "fig6_digital_twin_test_forecast.png")
plt.savefig(fig_path, dpi=300)
plt.show()
```

```python
import os
import json
import pickle
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# Directories
export_dir = "models"
os.makedirs(export_dir, exist_ok=True)

# 1. Load data & profile splits
df = pd.read_csv("pmsm_thermal_twin/data/processed/pmsm_featured.csv")
splits = np.load("pmsm_thermal_twin/data/processed/split_profiles.npz")

train_val_profiles = np.concatenate([splits['train_profiles'], splits['val_profiles']])
train_val_df = df[df['profile_id'].isin(train_val_profiles)].reset_index(drop=True)

ignore_cols = ['profile_id', 'stator_winding_future_60s', 'temp_rise_60s']
feature_cols = [c for c in df.columns if c not in ignore_cols]
target_col = 'stator_winding_future_60s'

X = train_val_df[feature_cols]
y = train_val_df[target_col]

# 2. Fit scaler and model
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = Ridge(alpha=100.0)
model.fit(X_scaled, y)

# 3. Export as .pkl format using standard pickle
with open(os.path.join(export_dir, "pmsm_ridge_model.pkl"), "wb") as f:
    pickle.dump(model, f)

with open(os.path.join(export_dir, "pmsm_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(export_dir, "feature_columns.json"), "w") as f:
    json.dump(feature_cols, f)

print("=== .PKL ARTIFACTS GENERATED SUCCESSFULLY ===")
print("Download these files from Colab to your local models/ folder:")
for item in os.listdir(export_dir):
    print(f"- {os.path.join(export_dir, item)}")
```
